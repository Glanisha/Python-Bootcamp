```python
%pip install openpyxl

# Import the pandas library
import pandas as pd

# --- Step 1: Load your Excel dataset ---
# Replace 'Orders.xlsx' with the actual name of your Excel file.
try:
    df = pd.read_excel('Orders.xlsx')
except FileNotFoundError:
    print("Error: 'Orders.xlsx' not found. Please make sure the file is in the same directory or provide the full path.")
    # Exiting if file not found, as we can't proceed.
    exit()

# --- Step 2: FIX and DEBUG - Print the exact column names ---
# This will show you the list of available columns so you can check for the exact name.
print("Columns found in your file:", df.columns)


# --- Step 3: Crucial Data Preparation ---
# Use the correct column name from the list printed above. It is likely 'Order Date'.
try:
    # Replace 'OrderDate' with the correct name if it's different.
    correct_date_column = 'OrderDate' # <-- CHANGE THIS to the name you see in the output above
    df[correct_date_column] = pd.to_datetime(df[correct_date_column], errors='coerce')

    # Display the first few rows and data types to confirm everything is loaded correctly
    print("\nDataset Head:")
    print(df.head())
    print("\nData Types:")
    df.info()

except KeyError:
    print(f"\nERROR: The column '{correct_date_column}' was not found. Please check the list of columns printed above and update the 'correct_date_column' variable in the code.")

```
