```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("exp4.csv")
df.head()
```

```python
df.info()
```

```python
df.drop(3, axis=0)
```

```python
df['Marks'].fillna(df['Marks'].mean(), inplace=True)
```

```python
df.head()
```

```python

df['Name'].value_counts().plot.pie()
plt.show()


df['Marks'].plot.hist()
plt.show()


df.groupby('Name')['Marks'].mean().plot.bar()
plt.show()


df['Marks'].plot.kde()
plt.show()
```
