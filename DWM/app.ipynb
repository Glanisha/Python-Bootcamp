```python
import pandas as pd 
```

```python
df= pd.read_excel('dwmexp3.xlsx')
```

```python
df.head()
```

```python
#roll up 

roll_up = df.groupby('Region')['Sales'].sum().reset_index()
print("ROLL-UP: Total Sales per Region")
print(roll_up)
```

```python
#drill down
drill_down = df.groupby(['Region', 'Product', 'Time'])['Sales'].sum().reset_index()
print("DRILL-DOWN: Sales per Region, Product, and Quarter")
print(drill_down)
```

```python
slice_df = df[df['Region'] == 'East']
print("SLICE: Only 'East' Region Data")
print(slice_df)
```

```python
dice_df = df[(df['Region'].isin(['East', 'North'])) & (df['Product'] == 'Laptop')]
print("DICE: East or North Regions AND Product = Laptop")
print(dice_df)
```
