```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np 
%matplotlib inline 
```

```python
dataset=pd.read_csv("Kaggle_Algerian_forest_fires_dataset.csv")
dataset.info()
```

```python
dataset.head()
#we will predict the temperature column
```

```python
dataset[dataset.isnull().any(axis=1)]
```

```python
#separating both the regions 
dataset.loc[:122, "Region"]=0
dataset.loc[122:, "Region"]=1
dataset.iloc[121]

```

```python
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.dropna(inplace=True)
```

```python
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset["Region"]=dataset["Region"].astype(int)
dataset.info()
```

```python
dataset.iloc[[122]]
```

```python
dataset = dataset.drop(dataset.index[122])
dataset.iloc[[122]]
```

```python
dataset.columns=dataset.columns.str.strip()
dataset.columns
```

```python
dataset.info()
```

```python
dataset[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']]=dataset[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)

dataset[['Rain', 'FFMC',
       'DMC', 'DC', 'ISI', 'BUI', 'FWI']]=dataset[['Rain', 'FFMC',
       'DMC', 'DC', 'ISI', 'BUI', 'FWI']].astype(float)
```

```python
dataset["Classes"].unique()
```

```python
dataset["Classes"]=dataset["Classes"].str.strip()
dataset["Classes"].unique()
```

```python
dataset.to_csv("Cleaned.csv")
```
